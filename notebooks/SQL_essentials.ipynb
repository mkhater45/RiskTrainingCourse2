```python
import sqlite3
import pandas as pd
import time

# 1. Connect to your local LI-Medium database
# (Ensure the file name matches your generated .sqlite file)
db_path = "bank_risk_aml_medium.sqlite"
conn = sqlite3.connect(db_path)
```

```python
# 2. Write the multi-table relational query
query = """
SELECT 
    t.timestamp,
    t.from_account,
    a.customer_type AS sender_type,
    t.to_account,
    t.amount_paid,
    t.payment_format,
    t.is_laundering
FROM transactions t
LEFT JOIN accounts a 
    ON t.from_account = a.account_id
WHERE t.is_laundering = 1
LIMIT 100;
"""

```

```python
# 3. Time and execute the query
start_time = time.time()
df_flagged = pd.read_sql_query(query, conn)
elapsed = time.time() - start_time

print(f" Query executed in {elapsed:.4f} seconds across millions of rows!\n")

# 4. Display the DataFrame in the notebook output
df_flagged
```
