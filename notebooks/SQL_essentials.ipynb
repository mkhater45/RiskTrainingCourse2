```python
import duckdb
import time
import pandas as pd

# Location of the data on a Microsoft Cloud server in .parquet format
TRANSACTIONS_TABLE ='https://tahastorage.blob.core.windows.net/training-public/transactions.parquet'
ACCOUNTS_TABLE ='https://tahastorage.blob.core.windows.net/training-public/accounts.parquet'

# Create a database connection object
conn = duckdb.connect()

# Optimize the network settings for the classroom
conn.execute("SET http_keep_alive = true;")
conn.execute("SET force_download = false;")
```

```python
# 2. Write the query
query = f"""

SELECT  *
FROM '{TRANSACTIONS_TABLE}'
limit 100
"""

```

```python
# 3. Time and execute the query
start_time = time.time()
results_frame = pd.read_sql_query(query, conn)
elapsed = time.time() - start_time

print(f" Query executed in {elapsed:.4f} seconds across millions of rows!\n")

# 4. Display the DataFrame in the notebook output
results_frame
```
